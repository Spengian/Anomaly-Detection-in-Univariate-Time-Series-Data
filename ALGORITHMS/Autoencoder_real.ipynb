## A1A

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
import matplotlib.pyplot as plt
from sklearn.metrics import precision_score, recall_score, f1_score

# Function to create dataset for AE
def create_dataset(data, look_back=24):
    X = []
    for i in range(len(data) - look_back):
        X.append(data[i:i + look_back])
    return np.array(X)

# Datasets
A1A = [1,2,3,8,9,10,13,23,24,27,34,41,66]

all_best_precisions_dw = []
all_best_recalls_dw = []
all_best_f1_scores_dw = []
all_best_precisions_sw = []
all_best_recalls_sw = []
all_best_f1_scores_sw = []

possible_ks = [1.5, 2, 2.5, 3]  # Διάφορες τιμές k για δοκιμή
possible_windows = [100, 200, 300, 500]  # Διάφορες τιμές rolling window

look_back = 24

for i in A1A:
    # Extract data and labels
    data = dfsA1[i][f'value_{i}'].values.astype(float)
    labels = dfsA1[i]['label'].values.astype(int)  # True anomaly labels (0 or 1)

    # Create dataset for AE (X only)
    X = create_dataset(data, look_back=look_back)

    # Align labels
    aligned_labels = labels[look_back:]

    # Split into train/test
    train_size = int(len(X) * 0.3)
    X_train, X_test = X[:train_size], X[train_size:]
    true_anomalies = aligned_labels[train_size:]  # True labels for test set

    # Scale data
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # Build a simple MLP Autoencoder
    model = Sequential()
    # Encoder
    model.add(Dense(32, activation='relu', input_shape=(look_back,)))
    model.add(Dense(16, activation='relu'))
    # Decoder
    model.add(Dense(16, activation='relu'))
    model.add(Dense(32, activation='relu'))
    model.add(Dense(look_back))  # Output layer reconstructing the original input

    model.compile(optimizer='adam', loss='mae')

    # Train the model
    history = model.fit(
        X_train_scaled,
        X_train_scaled,  # AE tries to reconstruct the input
        epochs=75,
        batch_size=32,
        verbose=0,
        validation_split=0.1
    )

    # Make reconstruction predictions on test data
    reconstructed = model.predict(X_test_scaled)

    # Inverse transform to compare errors in original scale
    X_test_inv = scaler.inverse_transform(X_test_scaled)
    reconstructed_inv = scaler.inverse_transform(reconstructed)

    # Calculate reconstruction errors
    errors = np.mean(np.abs(X_test_inv - reconstructed_inv), axis=1)

#=================================================================================================================
# TRUE AND PREDICTED AND LOSSES PLOTS
#=================================================================================================================

    # Plot training and validation loss
    plt.figure(figsize=(10, 6))
    plt.plot(history.history['loss'], label='Training Loss (MAE)')
    plt.plot(history.history['val_loss'], label='Validation Loss (MAE)')
    plt.title(f"Training and Validation Loss for Dataset {i} (MLP AE)")
    plt.xlabel('Epochs')
    plt.ylabel('Loss (MAE)')
    plt.legend()
    plt.show()

#=================================================================================================================
# RESIDUAL PLOT
#=================================================================================================================
#
#    plt.scatter(reconstructed_inv, errors, alpha=0.6)
#    plt.axhline(y=0, color='red', linestyle='--', linewidth=1)
#    plt.xlabel('Predicted Values')
#    plt.ylabel('Residuals')
#    plt.title('Residual Plot')
#    plt.show()
#
#=================================================================================================================
# STATIC WINDOW
#=================================================================================================================

    # Optimize k for the static threshold
    best_f1_sw = -1
    best_k_sw = None
    best_precision_sw = 0
    best_recall_sw = 0
    best_predicted_anomalies_sw = None

    for k_sw in possible_ks:
        threshold_sw = np.mean(errors) + k_sw * np.std(errors)
        predicted_anomalies_sw = (errors > threshold_sw).astype(int)

        precision_sw = precision_score(true_anomalies, predicted_anomalies_sw, zero_division=0)
        recall_sw = recall_score(true_anomalies, predicted_anomalies_sw, zero_division=0)
        f1_sw = f1_score(true_anomalies, predicted_anomalies_sw, zero_division=0)

        if f1_sw > best_f1_sw:
            best_f1_sw = f1_sw
            best_k_sw = k_sw
            best_precision_sw = precision_sw
            best_recall_sw = recall_sw
            best_predicted_anomalies_sw = predicted_anomalies_sw

    all_best_precisions_sw.append(best_precision_sw)
    all_best_recalls_sw.append(best_recall_sw)
    all_best_f1_scores_sw.append(best_f1_sw)

    print(f"Dataset {i}: Best k = {best_k_sw}")
    print(f" Precision={best_precision_sw:.2f}, Recall={best_recall_sw:.2f}, F1={best_f1_sw:.2f}")

    # Plot anomalies
    plt.figure(figsize=(10, 6))
    last_points = X_test_inv[:, -1]
    reconstructed_last_points = reconstructed_inv[:, -1]
    plt.plot(last_points, label='True Data (last timestep in window)')
    plt.plot(reconstructed_last_points, label='Reconstructed Data (last timestep in window)')
    anomaly_indices = np.where(best_predicted_anomalies_sw == 1)[0]
    plt.scatter(anomaly_indices, last_points[anomaly_indices], color='red', label='Anomalies')
    plt.legend()
    plt.title(f"Dataset {i}: Best k={best_k_sw} - Anomalies (Reconstruction Error)")
    plt.show()


#=================================================================================================================
# DYNAMIC WINDOW
#=================================================================================================================

    best_f1_for_dataset_dw = -1
    best_k_dw = None
    best_window_for_dataset_dw = None
    best_precision_for_dataset_dw = 0.0
    best_recall_for_dataset_dw = 0.0
    best_predicted_anomalies_dw = None

    for window_dw in possible_windows:
        # Υπολογισμός κυλιόμενων στατιστικών για το “dynamic threshold”
        rolling_residuals_dw = pd.Series(errors).rolling(window=window_dw, min_periods=1)
        mean_residuals_dw = rolling_residuals_dw.mean()
        std_residuals_dw = rolling_residuals_dw.std()

        for k_dw in possible_ks:
            # Ορισμός ορίου: mean + k * std
            threshold_dw = mean_residuals_dw + k_dw * std_residuals_dw
            predicted_anomalies_dw = (errors > threshold_dw).astype(int)

            precision_k_dw = precision_score(true_anomalies, predicted_anomalies_dw)
            recall_k_dw = recall_score(true_anomalies, predicted_anomalies_dw)
            f1_k_dw = f1_score(true_anomalies, predicted_anomalies_dw)

            if f1_k_dw > best_f1_for_dataset_dw:
                best_f1_for_dataset_dw = f1_k_dw
                best_k_dw = k_dw
                best_window_for_dataset_dw = window_dw
                best_precision_for_dataset_dw = precision_k_dw
                best_recall_for_dataset_dw = recall_k_dw
                best_predicted_anomalies_dw = predicted_anomalies_dw

    # Αποθήκευση καλύτερων τιμών για το συγκεκριμένο dataset
    all_best_precisions_dw.append(best_precision_for_dataset_dw)
    all_best_recalls_dw.append(best_recall_for_dataset_dw)
    all_best_f1_scores_dw.append(best_f1_for_dataset_dw)

    # Εμφάνιση καλύτερου αποτελέσματος
    print(f"Dataset {i} - Best k = {best_k_dw}, Best Dynamic Window = {best_window_for_dataset_dw}")
    print(f"Precision DW={best_precision_for_dataset_dw:.2f}, Recall DW={best_recall_for_dataset_dw:.2f}, F1 DW={best_f1_for_dataset_dw:.2f}")

    # Plot anomalies
    plt.figure(figsize=(10, 6))
    last_points = X_test_inv[:, -1]
    reconstructed_last_points = reconstructed_inv[:, -1]
    plt.plot(last_points, label='True Data (last timestep in window)')
    plt.plot(reconstructed_last_points, label='Reconstructed Data (last timestep in window)')
    anomaly_indices = np.where(best_predicted_anomalies_dw == 1)[0]
    plt.scatter(anomaly_indices, last_points[anomaly_indices], color='red', label='Anomalies')
    plt.legend()
    plt.title(f"Dataset {i}: Best k={best_k_dw} - Anomalies (Reconstruction Error)")
    plt.show()

#=================================================================================================================
# AVERAGE SCORES
#=================================================================================================================

average_precision_sw = np.mean(all_best_precisions_sw)
average_recall_sw = np.mean(all_best_recalls_sw)
average_f1_score_sw = np.mean(all_best_f1_scores_sw)

print("\n=== Average Best Metrics Across All Datasets ===")
print(f"Average Precision SW = {average_precision_sw:.2f}")
print(f"Average Recall SW   = {average_recall_sw:.2f}")
print(f"Average F1-Score SW = {average_f1_score_sw:.2f}")
formatted_scores_sw = [f"{score:.4f}" for score in all_best_f1_scores_sw]
print(f"All F1 Scores SW: {formatted_scores_sw}")


average_precision_dw = np.mean(all_best_precisions_dw)
average_recall_dw = np.mean(all_best_recalls_dw)
average_f1_score_dw = np.mean(all_best_f1_scores_dw)

print("\n=== Average Best Metrics Across All Datasets ===")
print(f"Average Precision DW = {average_precision_dw:.2f}")
print(f"Average Recall DW   = {average_recall_dw:.2f}")
print(f"Average F1-Score DW = {average_f1_score_dw:.2f}")
formatted_scores_dw = [f"{score:.4f}" for score in all_best_f1_scores_dw]
print(f"All F1 Scores DW: {formatted_scores_dw}")
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
import matplotlib.pyplot as plt
from sklearn.metrics import precision_score, recall_score, f1_score

# Function to create dataset for AE
def create_dataset(data, look_back=24):
    X = []
    for i in range(len(data) - look_back):
        X.append(data[i:i + look_back])
    return np.array(X)

look_back = 288

for file_name in data['artificialWithAnomaly']:
    print(f"Processing file: {file_name}...")

    data1 = data['artificialWithAnomaly'][file_name]['value'].values.astype(float)
    labels = data['artificialWithAnomaly'][file_name]['label'].values.astype(int)  # True anomaly labels (0 or 1)

    # Create dataset for AE (X only)
    X = create_dataset(data1, look_back=look_back)

    # Align labels
    aligned_labels = labels[look_back:]

    # Split into train/test
    train_size = int(len(X) * 0.4)
    X_train, X_test = X[:train_size], X[train_size:]
    true_anomalies = aligned_labels[train_size:]  # True labels for test set

    # Scale data
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # Build a simple MLP Autoencoder
    model = Sequential()
    # Encoder
    model.add(Dense(64, activation='relu', input_shape=(look_back,)))
    model.add(Dense(32, activation='relu'))
    # Decoder
    model.add(Dense(32, activation='relu'))
    model.add(Dense(64, activation='relu'))
    model.add(Dense(look_back))  # Output layer reconstructing the original input

    model.compile(optimizer='adam', loss='mae')

    # Train the model
    history = model.fit(
        X_train_scaled,
        X_train_scaled,  # AE tries to reconstruct the input
        epochs=75,
        batch_size=32,
        verbose=0,
        validation_split=0.1
    )

    # Make reconstruction predictions on test data
    reconstructed = model.predict(X_test_scaled)

    # Inverse transform to compare errors in original scale
    X_test_inv = scaler.inverse_transform(X_test_scaled)
    reconstructed_inv = scaler.inverse_transform(reconstructed)

    # Calculate reconstruction errors (mean absolute error per sample)
    errors = np.mean(np.abs(X_test_inv - reconstructed_inv), axis=1)

    # Determine anomalies using mean+3*std threshold
    mean_resid = np.mean(errors)
    std_resid = np.std(errors)
    threshold = mean_resid + 2.5 * std_resid

    predicted_anomalies = (errors > threshold).astype(int)

    # Offset to align detections with the global dataset index
   # offset = int(len(X) * 0.4)

    # Adjust detections by adding the offset
    detections = [i + look_back + train_size for i, x in enumerate(predicted_anomalies) if x == 1]
    print(f"Detections for {file_name}: {detections}")

    # Compute the NAB score
    nab_score = compute_raw_score_debug(detections, anomaly_windows[f"{file_name}"], A_tp = 1.0, A_fp = -0.11, A_fn = -1.0)

    # Compute metrics using sklearn
    precision = precision_score(true_anomalies, predicted_anomalies)
    recall = recall_score(true_anomalies, predicted_anomalies)
    f1 = f1_score(true_anomalies, predicted_anomalies)

    print(f"Dataset {file_name}")
    print(f" Precision={precision:.2f}, Recall={recall:.2f}")

    # Plot all true data with reconstruction and anomalies for the test set
    plt.figure(figsize=(14, 8))

    # Plot the entire true data series
    plt.plot(data1, label='True Data', alpha=0.7, color='blue')

    # Overlay reconstructed test data
    test_indices = np.arange(len(data1))[look_back + train_size:]
    plt.plot(test_indices, reconstructed_inv[:, -1], label='Reconstructed Data (Test)', color='orange')

    # Highlight anomalies
   # anomaly_indices = np.where(predicted_anomalies == 1)[0]
  #  plt.scatter(anomaly_indices, data1[anomaly_indices], color='red', label='Predicted Anomalies')
    plt.scatter(detections, data1[detections], color='red', label='Predicted Anomalies', marker='x')

    plt.legend()
    plt.title(f"Dataset {file_name}: All True Data and Reconstructed Anomalies")
    plt.xlabel("Timestamps")
    plt.ylabel("Value")
    plt.grid()
    plt.show()

    # Plot training and validation loss
    plt.figure(figsize=(10, 6))
    plt.plot(history.history['loss'], label='Training Loss (MAE)')
    plt.plot(history.history['val_loss'], label='Validation Loss (MAE)')
    plt.title(f"Training and Validation Loss for Dataset {file_name} (MLP AE)")
    plt.xlabel('Epochs')
    plt.ylabel('Loss (MAE)')
    plt.legend()
    plt.grid()
    plt.show()
```

## NAB Scoring function

```python
import math

def sigmoid(x):
    """Standard sigmoid function."""
    return 1 / (1 + math.exp(-x))

def sigmoid_score(t, A_tp, A_fp):
    """
    Calculate the scaled sigmoid score for a detection.
    @param t: Relative position of the detection:
              - t = -1 is far left of the window, t = 0 is the right edge.
              - t > 0 considers detections outside the window.
    @param A_tp: Reward for true positives.
    @param A_fp: Penalty for false positives.
    """
    if t > 3.0:
        return -1.0  # Assign -1 for detections too far after the window
    elif 0 < t <= 3.0:
        return  2* (1 / (1 + math.exp(5 * t))) - 1.0  # Sigmoid scaling
    else:
        return  2* (1 / (1 + math.exp(5 * t))) - 1.0  # Normal sigmoid for in-window cases

def compute_raw_score_debug(detections, anomaly_windows, A_tp, A_fp, A_fn):
    """
    Compute the raw NAB score for a dataset with detailed debugging.
    Also computes S_perfect and S_null for comparison.
    @param detections: List of detected anomaly timestamps.
    @param anomaly_windows: List of anomaly windows (start_time, end_time).
    @param A_tp: Reward for true positives.
    @param A_fp: Penalty for false positives.
    @param A_fn: Penalty for missed anomaly windows (false negatives).

    @return: Raw NAB score, S_perfect, S_null.
    """
    score = 0
    missed_windows = 0


    # Sort detections and anomaly windows
    detections = sorted(detections)
    anomaly_windows = sorted(anomaly_windows, key=lambda x: x[0])

    # Compute the actual score
    for i, window in enumerate(anomaly_windows):
        start_time, end_time = window
        window_length = end_time - start_time

        # Find detections within or after the current window
        detections_in_window = [d for d in detections if d >= start_time]

        if detections_in_window and start_time <= detections_in_window[0] <= end_time:
            # Detection within the window
            first_detection = detections_in_window[0]
            t = -3 + 2 * (first_detection - start_time) / window_length
            contribution = sigmoid_score(t, A_tp, A_fp)
            print(f"Window {i}: {window}, Detection t={t:.3f} at point {first_detection}, Contribution={contribution:.3f}")
            score += contribution
            # Remove the detection
            detections.remove(first_detection)
            detections_in_window = [d for d in detections if start_time <= d <= end_time]
            for d in detections_in_window:
                detections.remove(d)


        elif detections_in_window:
            # Check all detections just outside the current window
            near_miss_detections = [d for d in detections_in_window if 0 < (d - end_time) <= 3.0]

            for near_detection in near_miss_detections:
                t = (near_detection - end_time)  # Relative position after the window
                contribution = sigmoid_score(t, A_tp, A_fp)
                print(f"Window {i}: {window}, Near-miss detection t={t:.3f}, Contribution={contribution:.3f}")
                score += contribution
                # Remove the detection after scoring it
                detections.remove(near_detection)
        else:
            # No detections in this window - count as a false negative
            print(f"Window {i}: {window}, Missed anomaly window. Penalty={A_fn:.3f}")
            score += A_fn
            missed_windows += 1

    # Handle remaining detections as false positives or outside windows
    for detection in detections:
        # Check if detection is close to the end of any window
        near_miss_found = False
        for window in anomaly_windows:
            _, end_time = window
            if 0 < (detection - end_time) <= 3.0:
                t = (detection - end_time)
                contribution = sigmoid_score(t, A_tp, A_fp)
                print(f"Detection {detection} near-miss outside any window. t={t:.3f}, Contribution={contribution:.3f}")
                score += contribution
                near_miss_found = True
                break

        if not near_miss_found:
            # Treat as a false positive
            contribution = sigmoid_score(3.0, A_tp, A_fp)  # Use t=3.0 for far detections
            print(f"False positive detection at {detection}, Contribution={contribution:.3f}")
            score += contribution

    print(f"Raw score for dataset: {score:.3f}")

    return score
```

```python
def create_anomaly_windows(binary_labels):
    """
    Generate an array of anomaly windows (start, end) tuples from binary labels.

    @param binary_labels: List or array of binary labels (0s and 1s).
                          1 indicates an anomaly, 0 indicates normal.

    @return: List of tuples where each tuple is (start_index, end_index) for an anomaly window.
    """
    anomaly_windows = []
    in_window = False
    start_index = None

    for i, label in enumerate(binary_labels):
        if label == 1 and not in_window:
            # Start of a new anomaly window
            start_index = i
            in_window = True
        elif label == 0 and in_window:
            # End of the current anomaly window
            end_index = i - 1
            anomaly_windows.append((start_index, end_index))
            in_window = False

    # If the last label is part of an anomaly window, close it
    if in_window:
        anomaly_windows.append((start_index, len(binary_labels) - 1))

    return anomaly_windows
```
