```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import wilcoxon

# Αλγόριθμοι
algorithms = ["AE", "MLP", "LSTM", "XGBOOST", "LR"]

# Χρώματα
color_significant = '#2ca02c'  # πράσινο
color_not_significant = '#1f77b4'  # μπλε

fig, axs = plt.subplots(nrows=1, ncols=len(algorithms), figsize=(18, 6), sharey=True)

for i, algo in enumerate(algorithms):
    static_A1D = A1D[f"{algo}_static"]
    dynamic_A1D = A1D[f"{algo}_dynamic"]
    static_A1DD = A1DD[f"{algo}_static"]
    dynamic_A1DD = A1DD[f"{algo}_dynamic"]

    # Wilcoxon tests
    _, p_static = wilcoxon(static_A1D, static_A1DD)
    _, p_dynamic = wilcoxon(dynamic_A1D, dynamic_A1DD)

    data = [static_A1D, dynamic_A1D, static_A1DD, dynamic_A1DD]
    labels = ['A1D Static', 'A1D Dynamic', 'A1DD Static', 'A1DD Dynamic']

    # Χρώματα βάσει p-value
    colors = [
        color_significant if p_static < 0.05 else color_not_significant,
        color_significant if p_dynamic < 0.05 else color_not_significant,
        color_significant if p_static < 0.05 else color_not_significant,
        color_significant if p_dynamic < 0.05 else color_not_significant,
    ]

    bplot = axs[i].boxplot(data, patch_artist=True,
                           medianprops=dict(color='black'))

    # Set χρώματα κουτιών
    for patch, color in zip(bplot['boxes'], colors):
        patch.set_facecolor(color)

    axs[i].set_title(algo)
    axs[i].set_xticklabels(labels, rotation=45, fontsize=9)
    axs[i].grid(True, linestyle='--', alpha=0.5)

# Τίτλοι
fig.suptitle("Boxplots A1D vs A1DD για κάθε Αλγόριθμο (Static/Dynamic)", fontsize=16, fontweight="bold")
fig.supxlabel("Κατηγορίες", fontsize=12)
fig.supylabel("Τιμές", fontsize=12)

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
